# ✅ Colab 一键安装 Qiskit Aer GPU（CUDA 12.x）

本 Notebook 已为 **Google Colab (GPU)** 配置：
- 自动安装 **condacolab** 与 **mamba**；
- 将 **CUDA 12.5 运行库** 安装进同一 conda 环境；
- 使用 **PyPI** 安装 `qiskit-aer-gpu`（GPU 版 Aer），并避免被 CPU 版覆盖；
- 提供 **验证单元**，确保 `Device = GPU` 且可成功运行。

> 如果首次运行会触发一次 **内核重启**（由 `condacolab.install()` 引起）。重启后，从本单元继续往下运行即可。

```python
#@title 一键安装（Colab GPU）
import os, sys, subprocess, json, textwrap

print("Python:", sys.version)
print("Executable:", sys.executable)

# 0) 显示 GPU 可见性
print("=== NVIDIA SMI ===")
os.system("nvidia-smi || true")

# 1) 安装/激活 condacolab（可能会触发一次重启）
try:
    import condacolab  # noqa: F401
    print("condacolab already installed.")
except Exception:
    print("Installing condacolab...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "condacolab"])
    import condacolab
    condacolab.install()  # 这里会自动重启内核

# 2) 在 conda 环境中安装 CUDA 运行库（与系统驱动对接）
print("=== Installing CUDA runtime libs into conda env ===")
cuda_packages = [
    "cuda-version=12.5",
    "cuda-cudart=12.5.*",
    "libcublas=12.*",
    "libcusolver=11.*",
    "libcufft=11.*",
    "libcurand=10.*",
]
subprocess.check_call(["mamba", "install", "-y", "-c", "conda-forge"] + cuda_packages)

# 3) 确保未被 CPU 版 Aer 覆盖；安装 qiskit 与 GPU 版 Aer（PyPI 轮子）
print("=== Installing Qiskit + Aer GPU (PyPI) ===")
subprocess.call([sys.executable, "-m", "pip", "uninstall", "-y", "qiskit-aer"])  # remove CPU Aer if any
subprocess.check_call([sys.executable, "-m", "pip", "install", "-U", "qiskit"])
# CUDA 12 环境安装标准 GPU 包；若是 CUDA 11 可改为 qiskit-aer-gpu-cu11
subprocess.check_call([sys.executable, "-m", "pip", "install", "--no-cache-dir", "--force-reinstall", "qiskit-aer-gpu>=0.15"])

print("Setup complete.")
```

```python
#@title 验证：Aer 是否使用 GPU
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
import qiskit_aer as qa, sys
print("Python:", sys.version)
print("Qiskit-Aer version:", qa.__version__)

backend = Aer.get_backend("aer_simulator_statevector")
backend.set_options(device="GPU")
print("Device option:", backend.options.device)

# 简单电路测试
qc = QuantumCircuit(20)
qc.h(range(20))
qc.measure_all()
tqc = transpile(qc, backend)

res = backend.run(tqc, shots=256).result()
print("OK, sample:", list(res.get_counts().items())[:1])

# 进一步检查 CUDA 动态库可见性
import ctypes
missing = []
for lib in ["libcublas.so.12","libcusolver.so.11","libcufft.so.11","libcurand.so.10"]:
    try:
        ctypes.CDLL(lib)
        print(lib, "OK")
    except OSError as e:
        missing.append((lib, str(e)))
if missing:
    print("Missing CUDA libs:", missing)
```

## 3. Clone or refresh the QCNN repository
If the project directory already exists the cell will fetch the latest changes on the selected branch instead of cloning again.


```python
import os
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/NikolaStarx/QCNN_project.git"
BRANCH = "main"
PROJECT_DIR = Path("/content/QCNN_project")

if not PROJECT_DIR.exists():
    subprocess.run(["git", "clone", "-b", BRANCH, REPO_URL, str(PROJECT_DIR)], check=True)
else:
    subprocess.run(["git", "-C", str(PROJECT_DIR), "fetch", "origin"], check=True)
    subprocess.run(["git", "-C", str(PROJECT_DIR), "checkout", BRANCH], check=True)
    subprocess.run(["git", "-C", str(PROJECT_DIR), "pull", "origin", BRANCH], check=True)

os.chdir(PROJECT_DIR)
print(f"Working directory: {os.getcwd()}")
```

## 4. Install Python dependencies
Use pip to install the required packages listed in requirements.txt. This may take a few minutes the first time you run the notebook.


```python
USE_AER_GPU = True   # Set False to keep the CPU-only Aer build
AER_VERSION = "0.17.1"

import subprocess
import sys

def run(cmd):
    print("Running:", " ".join(cmd))
    subprocess.run(cmd, check=True)

run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"])

if USE_AER_GPU:
    run([sys.executable, "-m", "pip", "uninstall", "-y", "qiskit-aer"])
    run([sys.executable, "-m", "pip", "install", f"qiskit-aer-gpu"])
```

## 5. Configure the experiment
Select the YAML config you want to run and extract key preprocessing parameters. Update CONFIG_PATH to point to a different file if required.


```python
from pathlib import Path
import yaml

CONFIG_PATH = Path("configs/mnist_amplitude_quick.yaml")
with open(CONFIG_PATH, "r", encoding="utf-8") as stream:
    CONFIG = yaml.safe_load(stream)

DATASET = CONFIG["data"]["dataset"].lower()
ENCODING = CONFIG["data"]["encoding"]
NUM_QUBITS = CONFIG["data"]["num_qubits"]

print(f"Using config: {CONFIG_PATH}")
print(f"Dataset: {DATASET}, encoding: {ENCODING}, num_qubits: {NUM_QUBITS}")
```

## 6. Preprocess data
Download the dataset if needed and prepare the amplitude encoded tensors that the training script expects.


```python
import subprocess
import sys

preprocess_cmd = [
    sys.executable,
    "scripts/preprocess.py",
    DATASET,
    ENCODING,
    "--num_qubits",
    str(NUM_QUBITS),
]

print("Running:", " ".join(preprocess_cmd))
subprocess.run(preprocess_cmd, check=True)
```

## 7. Train the QCNN model
Launch the training loop using the selected configuration. Outputs appear below and checkpoints plus logs are saved under the project directory.


```python
import subprocess
import sys

train_cmd = [sys.executable, "train.py", "--config", str(CONFIG_PATH)]

print("Running:", " ".join(train_cmd))
subprocess.run(train_cmd, check=True)
```

## 8. Optional: package outputs
Copy logs and checkpoints into the output folder defined earlier and create a zip archive for download or Drive storage.


```python
import shutil
from datetime import datetime
from pathlib import Path

run_dir = OUTPUT_BASE / f"qcnn_run_{datetime.utcnow().strftime('%Y%m%d_%H%M%S')}"
run_dir.mkdir(parents=True, exist_ok=True)

for name in ["logs", "checkpoints"]:
    src = Path(name)
    if src.exists():
        dest = run_dir / name
        if dest.exists():
            shutil.rmtree(dest)
        shutil.copytree(src, dest)

archive_path = shutil.make_archive(str(run_dir), "zip", root_dir=run_dir)
archive_path
```
